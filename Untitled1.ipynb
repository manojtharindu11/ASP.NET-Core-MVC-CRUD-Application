```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

# We use Qwen2.5-1.5B because it is very efficient and has good multilingual support
# You can also try "google/gemma-2-2b-it"
model_name = "Qwen/Qwen2.5-1.5B-Instruct"

print(f"Loading {model_name}...")

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)

print("Model loaded successfully!")
```

```python
def ask_slm(question):
    # Format the message as a chat (User -> Assistant)
    messages = [
        {"role": "system", "content": "You are a helpful assistant for Sri Lankan History students. Answer in Sinhala."},
        {"role": "user", "content": question}
    ]

    # Prepare input
    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )
    model_inputs = tokenizer([text], return_tensors="pt").to(model.device)

    # Generate response
    generated_ids = model.generate(
        **model_inputs,
        max_new_tokens=512  # Limit the length of the answer
    )

    # Decode the output (remove the prompt part)
    generated_ids = [
        output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs.input_ids, generated_ids)
    ]
    response = tokenizer.batch_decode(generated_ids, skip_special_tokens=True)[0]

    return response

# --- TEST ---
# Question: Who is King Dutugemunu? (Written in Sinhala)
test_question = "දුටුගැමුණු රජතුමා යනු කවුද?"

print(f"Question: {test_question}\n")
print("Generating Answer... (This might take a few seconds)\n")
answer = ask_slm(test_question)

print("-" * 30)
print(f"SLM Answer:\n{answer}")
print("-" * 30)
```
